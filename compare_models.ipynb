# Model Comparison & Report
This notebook aggregates results from PyTorch and Keras experiments.

```python
import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from IPython.display import display

# Set style
sns.set_theme(style="whitegrid")
```

```python
# 1. Load Results
results_file = 'all_results.json'
if not os.path.exists(results_file):
    print("No results file found. Please run the training notebooks first.")
    data = {}
else:
    with open(results_file, 'r') as f:
        data = json.load(f)
        
print(f"Loaded results for tasks: {list(data.keys())}")
```

```python
# 2. Prepare Data for DataFrame
rows = []
for task, models in data.items():
    for model_name, metrics in models.items():
        row = {'Task': task, 'Model': model_name}
        # Flatten metrics
        for k, v in metrics.items():
            if isinstance(v, (int, float)):
                row[k] = v
            elif isinstance(v, dict):
                # Handle nested config if needed, or skip
                pass
        rows.append(row)

df = pd.DataFrame(rows)
print("Results Summary:")
display(df)
```

```python
# 3. Plot 1: Monk Test Accuracy
monk_df = df[df['Task'].str.contains('Monk')]
if not monk_df.empty:
    plt.figure(figsize=(10, 6))
    sns.barplot(data=monk_df, x='Task', y='test_accuracy', hue='Model', palette="viridis")
    plt.title("Monk Test Accuracy Comparison")
    plt.ylim(0, 1.1)
    plt.ylabel("Accuracy")
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.savefig("monk_comparison.png")
    plt.show()
else:
    print("No Monk results found.")
```

```python
# 4. Plot 2: CUP Internal Test MEE
cup_df = df[df['Task'] == 'CUP']
if not cup_df.empty:
    plt.figure(figsize=(8, 6))
    sns.barplot(data=cup_df, x='Task', y='internal_test_MEE', hue='Model', palette="magma")
    plt.title("CUP Internal Test MEE Comparison (Lower is Better)")
    plt.ylabel("Mean Euclidean Error (MEE)")
    plt.tight_layout()
    plt.savefig("cup_comparison.png")
    plt.show()
else:
    print("No CUP results found.")
```
